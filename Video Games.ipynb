```python
i = 'NA_Sales'

q1 = np.percentile(df[i],25)
q3 = np.percentile(df[i],75)

iqr = q3 - q1
```

```python
c1 = q1 - 1.5 * (iqr)
c2 = q3 + 1.5 * (iqr)
```

```python
t = df[ (df[i] > c2) | (df[i] < c1)]
```

```python
t = t.index
```

```python
df = df.drop(labels = t)
```
